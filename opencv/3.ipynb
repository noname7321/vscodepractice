```python
import cv2
import numpy as np
image = cv2.imread('saved_image.jpg')
```

```python
# 显示图像
cv2.imshow("Display Window", image)

# 等待按键输入
cv2.waitKey(0)

# 关闭所有窗口
cv2.destroyAllWindows()
```

```python
# 保存图像
cv2.imwrite("output_image.jpg", image)
```

```python
print(image.shape)  # 输出图像的尺寸 (高度, 宽度, 通道数)
```

```python
# 获取像素值 (BGR 格式)
pixel_value = image[100, 100]  # 获取 (100, 100) 处的像素值
print(f"Pixel value at (100, 100): {pixel_value}")
# 修改像素值
image[100, 100] = [255, 255, 255]  # 将 (100, 100) 处的像素设置为白色
print(f"Modified pixel value at (100, 100): {image[100, 100]}")
```

```python
# 获取 ROI
roi = image[50:150, 50:150]  # 获取 (50,50) 到 (150,150) 的区域
print(f"ROI shape: {roi.shape}")  # 输出 ROI 的尺寸
# 修改 ROI
image[50:150, 50:150] = [0, 255, 0]  # 将 ROI 区域设置为绿色
```

```python
cv2.imshow("Modified Image", image)
cv2.waitKey(0)
```

```python
# 分离通道
b, g, r = cv2.split(image)
print(f"Blue channel shape: {b.shape}, Green channel shape: {g.shape}, Red channel shape: {r.shape}")

# 合并通道
merged_image = cv2.merge([b, g, r])
print(f"Merged image shape: {merged_image.shape}")
```

```python
# 缩放
new_width = 200
new_height = 200
resized_image = cv2.resize(image, (new_width, new_height))#不会修改原始的image
cv2.imshow("Resized Image", resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
height, width = image.shape[:2]
center_x = width // 2
center_y = height // 2
angle = 45
scale = 1.0
```

```python
# 旋转
# (center_x, center_y)：旋转中心
# angle：旋转角度，单位是度
# scale：缩放比例
# 1.0：大小不变
# 0.5：缩小一半
# 2.0：放大两倍
rotation_matrix = cv2.getRotationMatrix2D((center_x, center_y), angle, scale)
#表示按照旋转矩阵生成新图像。最后的 (width, height) 是输出图像尺寸
rotated_image = cv2.warpAffine(image, rotation_matrix, (width, height))
cv2.imshow("Rotated Image", rotated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 平移
# tx：水平方向移动距离
# 正数：向右
# 负数：向左
# ty：垂直方向移动距离
# 正数：向下
# 负数：向上
tx = 100
ty = 50
translation_matrix = np.float32([[1, 0, tx], [0, 1, ty]])  # tx, ty 为平移距离
translated_image = cv2.warpAffine(image, translation_matrix, (width, height))
cv2.imshow("Translated Image", translated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 翻转\
#flip_code 决定翻转方向：
# 0: 垂直翻转
# 1: 水平翻转
# -1: 同时水平和垂直翻转
flip_code = 0  # 垂直翻转
flipped_image = cv2.flip(image, flip_code) 
cv2.imshow("Flipped Image", flipped_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
image1 = cv2.imread('saved_image.jpg')
image2 = cv2.imread('output_image.jpg')
result = cv2.add(image1, image2)
cv2.imshow("Added Image", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
result=cv2.subtract(image1, image2)
cv2.imshow("Subtracted Image", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
alpha = 0.5
beta = 0.5
gamma = 0
#各个参数的含义是：
# alpha: 第一个图像的权重
# beta: 第二个图像的权重
# gamma: 常数项，用于调整最终结果
result = cv2.addWeighted(image1, alpha, image2, beta, gamma)
cv2.imshow("Weighted Image", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#转化成灰度图像
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
#这段代码用于二值化图像，也就是把图像转换成主要只有黑色和白色的图像。
#其中，image是输入的图像，thresh是阈值，maxval是最大值，cv2.THRESH_BINARY是二值化的方法。
#大于阈值的像素值会被设置为最大值，小于阈值的像素值会被设置为0。
thresh = 127  # 阈值
maxval = 255  # 最大值
ret, thresholded_image = cv2.threshold(gray_image, thresh, maxval, cv2.THRESH_BINARY)
cv2.imshow("Thresholded Image", thresholded_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
block_size = 11
C = 2
# Adaptive thresholding
thresholded_image = cv2.adaptiveThreshold(gray_image, maxval, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, block_size, C)
cv2.imshow("Adaptive Thresholded Image", thresholded_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#使用 Otsu 自动计算阈值，并按照普通二值化规则生成黑白图像。
ret, thresholded_image = cv2.threshold(gray_image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print(f"Otsu's threshold value: {ret}")#会根据图片的直方图自动计算出一个最佳的阈值，
# 并将其存储在ret变量中。然后，使用这个阈值对图像进行二值化处理，
# 生成一个二值化后的图像thresholded_image。
# 最后，显示这个二值化后的图像，并等待用户按键后关闭窗口。
cv2.imshow("Otsu Thresholded Image", thresholded_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 使用一个 kernel_size × kernel_size5 的邻域处理每个像素。
# 然后计算平均值，并将其作为该像素的新值。
# kernel_size 是滤波器的大小，通常是一个奇数。
kernel_size = 5 
blurred_image = cv2.blur(image, (kernel_size, kernel_size))
cv2.imshow("Blurred Image", blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
sigmaX = 0
#sigmaX = 0时会根据 kernel_size 自动计算标准差，进行高斯模糊。
#高斯模糊会使用高斯分布的权重来计算每个像素的新值。
#通常来说中心像素的权重最大，周围像素的权重逐渐减小，这样可以有效地平滑图像，同时保留边缘信息。
blurred_image = cv2.GaussianBlur(image, (kernel_size, kernel_size), sigmaX)
cv2.imshow("Blurred Image", blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
blurred_image = cv2.medianBlur(image, kernel_size)#取中间值

#d: 邻域直径，表示每个像素的邻域大小（即参考周围像素的范围）。通常是一个奇数。
#sigmaColor: 控制颜色差异对滤波的影响，较小：只混合颜色非常接近的像素 较大：颜色差异较大的像素也会被混合，模糊更明显
#sigmaSpace: 控制空间距离的影响：较小：只影响附近像素 较大：更远的像素也可能参与计算
d = 9 #代表参考周围9*9的区域
sigmaColor = 75
sigmaSpace = 75
blurred_image = cv2.bilateralFilter(image, d, sigmaColor, sigmaSpace)
cv2.imshow("Blurred Image", blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# H：Hue，色调，表示颜色种类
# S：Saturation，饱和度，表示颜色鲜艳程度
# V：Value，明度，表示亮暗程度
hsv_img = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv_img)

cv2.imshow("H", h)
cv2.imshow("S", s)
cv2.imshow("V", v)

cv2.waitKey(0)
cv2.destroyAllWindows()

# Y：亮度
# U：蓝色色度差
# V：红色色度差
yuv_img = cv2.cvtColor(image, cv2.COLOR_BGR2YUV)
cv2.imshow("YUV Image", yuv_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
halfwidth = width // 2
halfheight = height // 2
resized_img = cv2.resize(image, (width // 2, height // 2))
cv2.imshow("Resized Image", resized_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
cropped_img = image[halfheight//2:halfheight//2 + halfheight, halfwidth//2:halfwidth//2 + halfwidth]
cv2.imshow("Cropped Image", cropped_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray_image, 100, 200)
cv2.imshow("Canny Edges", edges)
cv2.waitKey(0)
```

```python
# cv2.CV_64F：使用 64 位浮点数保存结果
# 1：dx=1，计算水平方向的变化
# 0：dy=0  计算垂直方向的变化
# ksize=5：使用 5 × 5 的 Sobel 核
sobel_x = cv2.Sobel(gray_image, cv2.CV_64F, 1, 0, ksize=5)
sobel_y = cv2.Sobel(gray_image, cv2.CV_64F, 0, 1, ksize=5)
abs_sobel_x = cv2.convertScaleAbs(sobel_x)
abs_sobel_y = cv2.convertScaleAbs(sobel_y)
cv2.imshow("Sobel X", abs_sobel_x)
cv2.imshow("Sobel Y", abs_sobel_y)
cv2.waitKey(0)
```

```python
sobel = cv2.addWeighted(
    abs_sobel_x,
    0.5,
    abs_sobel_y,
    0.5,
    0
)
cv2.imshow("Sobel Combined", sobel)
cv2.waitKey(0)
```

```python
laplacian = cv2.Laplacian(gray_image, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)
cv2.imshow("Laplacian", laplacian)
cv2.waitKey(0)
```

cv2.MORPH_RECT   # 矩形
cv2.MORPH_ELLIPSE  # 椭圆形
cv2.MORPH_CROSS  # 十字形

```python
_,binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
cv2.imshow("Binary Image", binary_image)
cv2.waitKey(0)
#这里创建了一个 5 × 5 的矩形核：
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
```

```python
#iterations=1 执行一次腐蚀
eroded_img = cv2.erode(binary_image, kernel, iterations=1)
cv2.imshow("Eroded Image", eroded_img)
cv2.waitKey(0)
```

```python
dilated_img = cv2.dilate(binary_image, kernel, iterations=1)
cv2.imshow("Dilated Image", dilated_img)
cv2.waitKey(0)
```

```python
# MORPH_OPEN：先腐蚀后膨胀，主要用于去除小的白色噪点。
# MORPH_CLOSE：先膨胀后腐蚀，主要用于填充小的黑色孔洞。
opening_img = cv2.morphologyEx(binary_image, cv2.MORPH_OPEN, kernel)
closing_img = cv2.morphologyEx(binary_image, cv2.MORPH_CLOSE, kernel)
cv2.imshow("Opening Image", opening_img)
cv2.waitKey(0)
cv2.imshow("Closing Image", closing_img)
cv2.waitKey(0)
```

cv2.RETR_LIST      # 查找所有轮廓，但不建立层级关系
cv2.RETR_EXTERNAL  # 只查找最外层轮廓
cv2.RETR_TREE      # 查找所有轮廓，并保存父子层级关系

cv2.CHAIN_APPROX_SIMPLE
表示压缩轮廓点。

例如一个矩形原本可能包含很多边缘点，但它通常只保存四个角点，从而节省内存。

如果使用：cv2.CHAIN_APPROX_NONE
则会保存轮廓边界上的所有点。

```python
_, threshold_img = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
# findContours() 实际返回两个结果：
# contours：轮廓坐标
# hierarchy：轮廓之间的层级关系
contours, _ = cv2.findContours(threshold_img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
# for contour in contours:
#     print(contour.shape)
```

```python
contour_image = image.copy()
# -1表示绘制所有轮廓。
cv2.drawContours(
    contour_image,#会直接修改原图
    contours,#轮廓
    -1,
    (0, 255, 0),#绿色
    1#线条粗细，-1会直接填充轮廓内部
)

cv2.imshow("Contours", contour_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
