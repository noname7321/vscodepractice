```python
import cv2

# 读取图像
img = cv2.imread('saved_image.jpg')

# 访问像素值
pixel_value = img[100, 150]  # 访问 (150, 100) 位置的像素值
print(pixel_value)
```

```python
# 修改像素值
img[100, 150] = [255, 255, 255]  # 将 (150, 100) 位置的像素值设置为白色
```

```python
# 我们想要提取图像中左上角 100x100 像素的区域：
# 提取 ROI
roi = img[0:100, 0:100]
```

```python
# 修改 ROI
roi[:, :] = [0, 255, 0]  # 将 ROI 区域设置为绿色,会直接修改img中的对应区域

# 将修改后的 ROI 放回原图像
img[0:100, 0:100] = roi
cv2.imwrite('modified_image.jpg', img)  # 保存修改后的图像
```

```python
# 通道分离，可以使用 cv2.split() 函数将图像的通道分离。
b, g, r = cv2.split(img)
```

```python
# 通道合并，分离后的通道可以使用 cv2.merge() 函数合并回原图像。
merged_img = cv2.merge([b, g, r])
```

```python
# 图像缩放
resized_img = cv2.resize(img, (200, 200))  # 将图像缩放为 200x200 像素
```

```python
# 图像旋转
(h, w) = img.shape[:2]# (h,w,c) = img.shape
center = (w // 2, h // 2)
M = cv2.getRotationMatrix2D(center, 45, 1.0)  # 旋转 45 度，1.0 表示不缩放，如果是 0.5 就是缩小一半，2.0 就是放大一倍
rotated_img = cv2.warpAffine(img, M, (w, h))    #这里用的是w，h，即新图像的宽度和高度
cv2.imwrite('rotated_image.jpg', rotated_img)  # 保存旋转后的图像
```

```python
import numpy as np
# 图像平移
M = np.float32([[1, 0, 100], [0, 1, 50]])  # 向右平移 100 像素，向下平移 50 像素
translated_img = cv2.warpAffine(img, M, (w, h)) #应用矩阵效果
```

```python
# 图像翻转
flipped_img = cv2.flip(img, 1)  # 水平翻转
cv2.imshow('Flipped Image', flipped_img)
cv2.waitKey(0)
```

```python
filpped_img_vertical = cv2.flip(img, 0)  # 垂直翻转
cv2.imshow('original Image', img)
cv2.imshow('Flipped Image Vertical', filpped_img_vertical)
cv2.waitKey(0)
```
