```python
import numpy as np

np.random.seed(0)

# 1. 造数据
m = 200
X = np.random.randn(2, m)          # shape: (2, m)，里面的数是从标准正态分布随机采样的：
Y = (X[0, :] + X[1, :] > 0).astype(float).reshape(1, m)  # shape: (1, m)，标签是根据X的两行之和是否大于0来生成的，满足条件的标签为1，不满足的标签为0。
# 取第0行和第1行的和，如果大于0，则标签为true，否则为false。最后将标签转换为浮点数，并调整形状为(1, m)。
# 2. 初始化参数
w = np.random.randn(2, 1) * 0.01
b = 0.0

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

lr = 0.1
num_iters = 6000

for i in range(num_iters):
    # 3. forward
    Z = w.T @ X + b                # shape: (1, m)
    A = sigmoid(Z)                 # shape: (1, m)

    # 4. loss
    eps = 1e-8
    loss = -np.mean(Y * np.log(A + eps) + (1 - Y) * np.log(1 - A + eps))

    # 5. backward
    dZ = A - Y                     # shape: (1, m)
    dw = (1 / m) * X @ dZ.T        # shape: (2, 1)
    db = (1 / m) * np.sum(dZ)

    # 6. update
    w = w - lr * dw
    b = b - lr * db

    if i % 100 == 0:
        pred = (A > 0.5).astype(float)
        acc = np.mean(pred == Y)
        print(f"iter {i:4d} | loss={loss:.4f} | acc={acc:.4f}")

print("w =", w.ravel())
print("b =", b)
```

```python
np.random.seed(0)

# 1. 造数据
m = 200
X = np.random.randn(2, m)          # shape: (2, m)，里面的数是从标准正态分布随机采样的：
Y = (X[0, :] + X[1, :] > 0).astype(float).reshape(1, m)  # shape: (1, m)，标签是根据X的两行之和是否大于0来生成的，满足条件的标签为1，不满足的标签为0。
# 取第0行和第1行的和，如果大于0，则标签为true，否则为false。最后将标签转换为浮点数，并调整形状为(1, m)。
# 2. 初始化参数
w1 = np.random.randn(2, 2) * 0.01
w2 = np.random.randn(2, 1) * 0.01
b1 = 0.0
b2 = 0.0

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

lr = 0.1
num_iters = 6000

for i in range(num_iters):
    # 3. forward
    Z1 = w1.T @ X + b1                # shape: (2, m)
    A1 = sigmoid(Z1)                 # shape: (2, m)
    Z2 = w2.T @ A1 + b2               # shape: (1, m)
    A2 = sigmoid(Z2)                  # shape: (1, m)

    # 4. loss
    eps = 1e-8
    loss = -np.mean(Y * np.log(A2 + eps) + (1 - Y) * np.log(1 - A2 + eps))

    # 5. backward
    dZ2 = A2 - Y                     # shape: (1, m)
    dw2 = (1 / m) * A1 @ dZ2.T        # shape: (2, 1)
    db2 = (1 / m) * np.sum(dZ2)

    dZ1 = w2 @ dZ2 * A1 * (1 - A1)   # shape: (2, m)
    dw1 = (1 / m) * X @ dZ1.T        # shape: (2, 2)
    db1 = (1 / m) * np.sum(dZ1, axis=1, keepdims=True)

    # 6. update
    w1 = w1 - lr * dw1
    w2 = w2 - lr * dw2
    b1 = b1 - lr * db1
    b2 = b2 - lr * db2

    if i % 100 == 0:
        pred = (A2 > 0.5).astype(float)
        acc = np.mean(pred == Y)
        print(f"iter {i:4d} | loss={loss:.4f} | acc={acc:.4f}")

print("w1 =", w1.ravel())
print("w2 =", w2.ravel())
print("b1 =", b1)
print("b2 =", b2)
```
