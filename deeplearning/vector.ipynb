```python
import numpy as np
a=np.random.rand(5)
print(a)#这里的a是一个长度为5的一维数组，里面的元素是随机生成的浮点数，范围在0到1之间。
print(a.shape)#这里的shape属性返回了a的形状，即(5,)，表示a是一个长度为5的一维数组。
print(a.T)#这里的T属性返回了a的转置，但是由于a是一维数组，所以转置后还是一个长度为5的一维数组。
#它既不是行向量也不是列向量，而是一个一维数组。它没有明确的行数和列数，只有一个维度。
#如果你想把它变成一个行向量，可以使用reshape方法：
print(np.dot(a,a.T))#这里的dot函数计算了a和a的转置的点积，结果是一个标量。
a = a.reshape(5, 1)
print(a)#这里的a现在是一个5x1的二维数组，也就是一个列向量。
print(np.dot(a,a.T))#这里的dot函数计算了a和a的转置的点积，结果是一个5x5的矩阵。
```
