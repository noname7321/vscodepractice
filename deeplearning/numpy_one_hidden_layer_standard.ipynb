# NumPy One-Hidden-Layer Neural Network

This notebook uses the same matrix convention as the neural-network formulas:

$$
Z^{[1]} = W^{[1]}X + b^{[1]}
$$

$$
A^{[1]} = g(Z^{[1]})
$$

$$
Z^{[2]} = W^{[2]}A^{[1]} + b^{[2]}
$$

$$
A^{[2]} = \sigma(Z^{[2]})
$$

Each column of `X` is one sample.


```python
import numpy as np

np.random.seed(0)
```

## 1. Create Toy Data

The label rule is:

$$
y = 1 \quad \text{if } x_1 + x_2 > 0
$$

Otherwise:

$$
y = 0
$$


```python
m = 200
X = np.random.randn(2, m)  # shape: (2, m); each column is one sample
Y = (X[0, :] + X[1, :] > 0).astype(float).reshape(1, m)  # shape: (1, m)

print("X.shape =", X.shape)
print("Y.shape =", Y.shape)
print("first 5 labels =", Y[:, :5])
```

## 2. Initialize Parameters

Use standard neural-network shapes:

- `W1`: `(n_h, n_x)`
- `b1`: `(n_h, 1)`
- `W2`: `(n_y, n_h)`
- `b2`: `(n_y, 1)`


```python
n_x = 2
n_h = 2
n_y = 1

W1 = np.random.randn(n_h, n_x) * 0.01
b1 = np.zeros((n_h, 1))
W2 = np.random.randn(n_y, n_h) * 0.01
b2 = np.zeros((n_y, 1))

print("W1.shape =", W1.shape)
print("b1.shape =", b1.shape)
print("W2.shape =", W2.shape)
print("b2.shape =", b2.shape)
```

## 3. Helper Function


```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

## 4. Train the Network

Forward propagation:

$$
Z^{[1]} = W^{[1]}X + b^{[1]}
$$

$$
A^{[1]} = \sigma(Z^{[1]})
$$

$$
Z^{[2]} = W^{[2]}A^{[1]} + b^{[2]}
$$

$$
A^{[2]} = \sigma(Z^{[2]})
$$

Backward propagation:

$$
dZ^{[2]} = A^{[2]} - Y
$$

$$
dW^{[2]} = \frac{1}{m} dZ^{[2]} (A^{[1]})^T
$$

$$
dZ^{[1]} = ((W^{[2]})^T dZ^{[2]}) \odot A^{[1]}(1-A^{[1]})
$$

$$
dW^{[1]} = \frac{1}{m} dZ^{[1]} X^T
$$


```python
lr = 0.1
num_iters = 6000

for i in range(num_iters):
    # Forward propagation
    Z1 = W1 @ X + b1       # shape: (n_h, m)
    A1 = sigmoid(Z1)       # shape: (n_h, m)
    Z2 = W2 @ A1 + b2      # shape: (n_y, m)
    A2 = sigmoid(Z2)       # shape: (n_y, m)

    # Loss
    eps = 1e-8
    loss = -np.mean(Y * np.log(A2 + eps) + (1 - Y) * np.log(1 - A2 + eps))

    # Backward propagation
    dZ2 = A2 - Y                              # shape: (n_y, m)
    dW2 = (1 / m) * dZ2 @ A1.T                # shape: (n_y, n_h)
    db2 = (1 / m) * np.sum(dZ2, axis=1, keepdims=True)

    dZ1 = (W2.T @ dZ2) * A1 * (1 - A1)        # shape: (n_h, m)
    dW1 = (1 / m) * dZ1 @ X.T                 # shape: (n_h, n_x)
    db1 = (1 / m) * np.sum(dZ1, axis=1, keepdims=True)

    # Gradient descent update
    W1 = W1 - lr * dW1
    b1 = b1 - lr * db1
    W2 = W2 - lr * dW2
    b2 = b2 - lr * db2

    if i % 500 == 0:
        pred = (A2 > 0.5).astype(float)
        acc = np.mean(pred == Y)
        print(f"iter {i:4d} | loss={loss:.4f} | acc={acc:.4f}")

pred = (A2 > 0.5).astype(float)
acc = np.mean(pred == Y)
print("final loss =", loss)
print("final acc =", acc)
print("W1 =\n", W1)
print("b1 =", b1.ravel())
print("W2 =\n", W2)
print("b2 =", b2.ravel())
```

## 5. Shape Checklist

If the formulas are correct, these shapes should hold:

- `Z1`, `A1`, `dZ1`: `(n_h, m)`
- `Z2`, `A2`, `dZ2`: `(n_y, m)`
- `dW1`: same shape as `W1`
- `db1`: same shape as `b1`
- `dW2`: same shape as `W2`
- `db2`: same shape as `b2`


```python
print("Z1.shape =", Z1.shape)
print("A1.shape =", A1.shape)
print("dZ1.shape =", dZ1.shape)
print("W1.shape =", W1.shape, "dW1.shape =", dW1.shape)
print("b1.shape =", b1.shape, "db1.shape =", db1.shape)
print("Z2.shape =", Z2.shape)
print("A2.shape =", A2.shape)
print("dZ2.shape =", dZ2.shape)
print("W2.shape =", W2.shape, "dW2.shape =", dW2.shape)
print("b2.shape =", b2.shape, "db2.shape =", db2.shape)
```
